# Benchmark Forecasting Models

## Appliance Energy Forecasting

This notebook establishes a set of simple forecasting benchmarks for hourly appliance energy use.

The benchmark models are:

- Mean forecast
- Naive forecast
- Daily seasonal naive forecast
- Weekly seasonal naive forecast
- Drift forecast

The final 14 days of the hourly dataset are reserved as the test period. Because the data are hourly, a daily seasonal period corresponds to 24 observations and a weekly seasonal period corresponds to 168 observations.

All benchmark forecasts are evaluated using the same accuracy metrics:

- Mean Absolute Error (MAE)
- Root Mean Squared Error (RMSE)
- Mean Absolute Scaled Error (MASE)
- Forecast Bias

The strongest benchmark will provide the reference against which the more complex SARIMAX, feature-based, and foundation models are evaluated.

## Forecasting Design

The modelling target is hourly appliance energy use (`Appliances`).

The forecasting experiment uses a chronological train-test split rather than a random split because future observations must not be used to predict the past.

The final 14 days are reserved for testing:

- Test period: 14 days
- Observations per day: 24
- Test observations: 336

The principal forecasting horizon is 24 hours.

For hourly observations:

- Daily seasonal period = 24
- Weekly seasonal period = 168

All benchmark models are evaluated on exactly the same held-out observations to ensure a fair comparison.

##  Evaluation Metrics

Four common metrics are used consistently across all forecasting models.

**MAE** measures the average absolute prediction error and is expressed in the same units as the target.

**RMSE** gives greater weight to large forecast errors and is therefore useful for detecting models that perform poorly during high-consumption events.

**MASE** scales forecast error relative to an in-sample seasonal naive benchmark, enabling meaningful comparison across forecasting methods.

**Bias** measures the average signed forecast error and indicates whether predictions systematically overestimate or underestimate appliance use.

```python
# RMSE

def calculate_rmse(y_true, y_pred):
    """
    Calculate root mean squared error.
    """
    return np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    mean_squared_error(..., squared=False)
```

```python
# MASE

def calculate_mase(
    y_true,
    y_pred,
    y_train,
    seasonality=24
):
    """
    Calculate Mean Absolute Scaled Error (MASE).

    The scaling denominator is based on the
    in-sample seasonal naive forecast error.
    """

    y_true = np.asarray(
        y_true,
        dtype=float
    )

    y_pred = np.asarray(
        y_pred,
        dtype=float
    )

    y_train = np.asarray(
        y_train,
        dtype=float
    )

    seasonal_error = np.abs(
        y_train[seasonality:]
        - y_train[:-seasonality]
    )

    scale = seasonal_error.mean()

    if scale == 0:
        return np.nan

    mae = np.mean(
        np.abs(
            y_true - y_pred
        )
    )

    return mae / scale
```

```python
# Common evaluation function

def evaluate_forecast(
    model_name,
    y_true,
    y_pred,
    y_train
):
    """
    Evaluate a forecasting model using
    MAE, RMSE, MASE, and Bias.
    """

    y_true = pd.Series(
        y_true
    ).astype(float)

    y_pred = pd.Series(
        y_pred,
        index=y_true.index
    ).astype(float)

    return {
        "Model": model_name,

        "MAE": mean_absolute_error(
            y_true,
            y_pred
        ),

        "RMSE": calculate_rmse(
            y_true,
            y_pred
        ),

        "MASE": calculate_mase(
            y_true,
            y_pred,
            y_train,
            seasonality=DAILY_PERIOD
        ),

        "Bias": np.mean(
            y_pred.values
            - y_true.values
        )
    }
```

##  Benchmark Forecasting Models

Simple benchmark models provide essential reference points for assessing whether more complex forecasting approaches add genuine predictive value.

Five benchmark methods are evaluated:

1. Mean forecast
2. Naive forecast
3. Daily seasonal naive forecast
4. Weekly seasonal naive forecast
5. Drift forecast

```python
# Mean forecast

def mean_forecast(
    y_train,
    horizon,
    forecast_index
):
    """
    Forecast all future observations using
    the mean of the training series.
    """

    return pd.Series(
        y_train.mean(),
        index=forecast_index,
        name="Mean"
    )
```

```python
# Naive forecast

def naive_forecast(
    y_train,
    horizon,
    forecast_index
):
    """
    Forecast all future observations using
    the final observed training value.
    """

    return pd.Series(
        y_train.iloc[-1],
        index=forecast_index,
        name="Naive"
    )
```

```python
# Seasonal naive forecast

def seasonal_naive_forecast(
    y_train,
    horizon,
    forecast_index,
    seasonality
):
    """
    Generate a recursive seasonal naive forecast.

    Each forecast uses the observation from
    the corresponding position in the previous
    seasonal cycle.
    """

    history = list(
        y_train.values
    )

    forecasts = []

    for _ in range(horizon):

        seasonal_value = (
            history[-seasonality]
        )

        forecasts.append(
            seasonal_value
        )

        history.append(
            seasonal_value
        )

    return pd.Series(
        forecasts,
        index=forecast_index
    )
```

```python
#  Drift forecast

def drift_forecast(
    y_train,
    horizon,
    forecast_index
):
    """
    Generate a drift forecast based on
    the average change between the first
    and final training observations.
    """

    slope = (
        y_train.iloc[-1]
        - y_train.iloc[0]
    ) / (
        len(y_train) - 1
    )

    forecast_values = [
        y_train.iloc[-1]
        + slope * step
        for step in range(
            1,
            horizon + 1
        )
    ]

    return pd.Series(
        forecast_values,
        index=forecast_index,
        name="Drift"
    )
```

```python
# Generate benchmark forecasts

benchmark_forecasts = {}

benchmark_forecasts["Mean"] = (
    mean_forecast(
        y_train=train,
        horizon=len(test),
        forecast_index=test.index
    )
)

benchmark_forecasts["Naive"] = (
    naive_forecast(
        y_train=train,
        horizon=len(test),
        forecast_index=test.index
    )
)

benchmark_forecasts[
    "Daily Seasonal Naive"
] = seasonal_naive_forecast(
    y_train=train,
    horizon=len(test),
    forecast_index=test.index,
    seasonality=DAILY_PERIOD
)

benchmark_forecasts[
    "Weekly Seasonal Naive"
] = seasonal_naive_forecast(
    y_train=train,
    horizon=len(test),
    forecast_index=test.index,
    seasonality=WEEKLY_PERIOD
)

benchmark_forecasts["Drift"] = (
    drift_forecast(
        y_train=train,
        horizon=len(test),
        forecast_index=test.index
    )
)

print(
    "Number of benchmark models:",
    len(benchmark_forecasts)
)

for name, forecast in benchmark_forecasts.items():
    print(
        name,
        "->",
        len(forecast),
        "predictions"
    )
```

```python
# Build forecast table

benchmark_forecast_df = pd.DataFrame(
    {
        "Actual": test
    }
)

for name, forecast in benchmark_forecasts.items():
    benchmark_forecast_df[name] = (
        forecast
    )

benchmark_forecast_df.head(10)
```

```python
# Evaluate all benchmarks

benchmark_results = []

for (
    model_name,
    forecast
) in benchmark_forecasts.items():

    result = evaluate_forecast(
        model_name=model_name,
        y_true=test,
        y_pred=forecast,
        y_train=train
    )

    benchmark_results.append(
        result
    )


benchmark_results_df = (
    pd.DataFrame(
        benchmark_results
    )
    .sort_values(
        by="RMSE"
    )
    .reset_index(
        drop=True
    )
)

benchmark_results_df.round(3)
```

##  Rolling-Origin 24-Hour Evaluation

The assignment defines a 24-hour forecasting horizon while reserving the final 14 days for testing.

To evaluate performance across the full test period without extending a single forecast to 336 hours, a rolling-origin evaluation is used.

At each daily forecast origin, the models use all observations available up to that point and generate the next 24 hourly predictions. Once the actual observations for that day become available, the training history is expanded and the next 24-hour forecast is generated.

This produces 14 non-overlapping 24-hour forecast windows covering the complete 14-day test period while preserving chronological order and preventing future observations from being used prematurely.

```python
def rolling_benchmark_evaluation(
    full_series,
    initial_train_size,
    test_steps,
    horizon=24
):
    """
    Generate rolling-origin benchmark forecasts.

    At each forecast origin:
    - use all observations available up to that point;
    - forecast the next 24 hours;
    - advance by 24 hours;
    - repeat until the complete test period is covered.
    """

    rolling_forecasts = {
        "Mean": [],
        "Naive": [],
        "Daily Seasonal Naive": [],
        "Weekly Seasonal Naive": [],
        "Drift": []
    }

    actual_values = []
    forecast_indices = []

    number_of_origins = test_steps // horizon

    for origin in range(number_of_origins):

        train_end = (
            initial_train_size
            + origin * horizon
        )

        forecast_end = (
            train_end
            + horizon
        )

        history = full_series.iloc[:train_end]

        actual_window = full_series.iloc[
            train_end:forecast_end
        ]

        forecast_index = actual_window.index

        rolling_forecasts["Mean"].append(
            mean_forecast(
                history,
                horizon,
                forecast_index
            )
        )

        rolling_forecasts["Naive"].append(
            naive_forecast(
                history,
                horizon,
                forecast_index
            )
        )

        rolling_forecasts[
            "Daily Seasonal Naive"
        ].append(
            seasonal_naive_forecast(
                history,
                horizon,
                forecast_index,
                DAILY_PERIOD
            )
        )

        rolling_forecasts[
            "Weekly Seasonal Naive"
        ].append(
            seasonal_naive_forecast(
                history,
                horizon,
                forecast_index,
                WEEKLY_PERIOD
            )
        )

        rolling_forecasts["Drift"].append(
            drift_forecast(
                history,
                horizon,
                forecast_index
            )
        )

        actual_values.append(actual_window)
        forecast_indices.extend(forecast_index)

    actual_series = pd.concat(
        actual_values
    )

    combined_forecasts = {}

    for name, forecasts in rolling_forecasts.items():
        combined_forecasts[name] = pd.concat(
            forecasts
        )

    return actual_series, combined_forecasts
```

```python
INITIAL_TRAIN_SIZE = len(target) - TEST_STEPS

rolling_actual, rolling_forecasts = (
    rolling_benchmark_evaluation(
        full_series=target,
        initial_train_size=INITIAL_TRAIN_SIZE,
        test_steps=TEST_STEPS,
        horizon=FORECAST_HORIZON
    )
)

print(
    "Actual observations:",
    len(rolling_actual)
)

for name, forecast in rolling_forecasts.items():
    print(
        name,
        "->",
        len(forecast)
    )
```

```python
# Evaluate rolling forecasts

rolling_results = []

for model_name, forecast in rolling_forecasts.items():

    result = evaluate_forecast(
        model_name=model_name,
        y_true=rolling_actual,
        y_pred=forecast,
        y_train=train
    )

    rolling_results.append(result)


rolling_results_df = (
    pd.DataFrame(rolling_results)
    .sort_values(
        by="RMSE"
    )
    .reset_index(drop=True)
)

rolling_results_df.round(3)
```
